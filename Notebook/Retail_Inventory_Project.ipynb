
# Retail Inventory Analysis using Pandas

This notebook contains only Pandas-based data cleaning and analysis.


```
import pandas as pd
```

## Load Dataset

```
df = pd.read_csv('retail_store_inventory(1).csv')

df.head()
```

## Shape of Dataset

```
df.shape
```

## Column Names

```
df.columns
```

## Dataset Information

```
df.info()
```

## Check Missing Values

```
df.isnull().sum()
```

## Check Duplicate Rows

```
df.duplicated().sum()
```

## Remove Duplicate Rows

```
df = df.drop_duplicates()

df.duplicated().sum()
```

## Check Data Types

```
df.dtypes
```

## Statistical Summary

```
df.describe()
```

## Unique Values in Category Column

```
df['Category'].unique()
```

## Total Units Sold by Category

```
df.groupby('Category')['Units Sold'].sum().sort_values(ascending=False)
```

## Average Price by Category

```
df.groupby('Category')['Price'].mean().sort_values(ascending=False)
```

## Inventory Level by Region

```
df.groupby('Region')['Inventory Level'].mean().sort_values(ascending=False)
```

## Units Sold by Seasonality

```
df.groupby('Seasonality')['Units Sold'].sum().sort_values(ascending=False)
```

## Average Units Sold by Weather Condition

```
df.groupby('Weather Condition')['Units Sold'].mean().sort_values(ascending=False)
```

## Average Units Sold by Promotion

```
df.groupby('Promotion')['Units Sold'].mean()
```

## Average Units Sold by Discount

```
df.groupby('Discount')['Units Sold'].mean().sort_values(ascending=False)
```

## Export Cleaned Dataset

```
df.to_csv('retail_store_inventory_cleaned.csv', index=False)

print('Dataset exported successfully')
```
